```python
import pandas as pd
```

```python
df = pd.read_csv("Titanic dataset.csv")
```

```python
df
```

```python
# Filtering passengers who are adults (Age >= 18) and Fare > 30
filtered_df = df[(df["Age"] >= 18) & (df['Fare'] > 30)]
```

```python
filtered_df
```

```python
# Creating new features
df["Family_Size"] = df['sibsp'] + df['Parch'] + 1
df["Fare_Per_Person"] = df['Fare']/df['Family_Size']
```

```python
df
```

```python
df[["Family_Size","Fare_Per_Person"]].head()
```

```python
# Average fare based on passenger class and gender
Avg_fare = df.groupby(["Pclass","Sex"])["Fare"].mean()
Avg_fare
```

```python
df.rename(columns={'2urvived': 'Survived'}, inplace=True)
```
