```python
# Write a program to find the largest of three numbers
a,b,c = map(int,input("Enter any three numbers : ").split())
if a>b and a>c:
    d = a
if b>a and b>c:
    d = b
if c>a and c>b:
    d = c
print("The largest number is ",d)
```

```python
# Write a program to check whether a given year is a leap year.
y = int(input("Enter a year : "))
if y%100 == 0:
    if y%400 == 0:
        print("It is a leap year")
    else :
        print("It is an ordinary year")
if y%100 != 0:
    if y%4 ==0:
        print("It ia a leap year")
    else :
        print("It is an ordinary year")
```

```python
# Write a program to check if a character is a vowel or a consonant.
w = input("Enter a letter : ").lower()
if w in ['a','e','i','o','u']:
    print("It is a vowel")
else: print("It is a consonant")
```

```python
# Write a program to calculate the sum of the first N natural numbers using a while loop.
n = int(input("Enter no of terms : "))
a=0
d=n
while n>0:
    a=a+n
    n-=1
print(f'The sum of first {d} natural numbers is {a}')
```

```python
# Write a program to print the multiplication table of a given number using a while loop.
n = int(input("Enter the number : "))
a=1
while a<=20:
    print(f'{n} x {a} = {n*a}')
    a+=1
```

```python
# pattern
n = int(input("Enter no of rows : "))
for i in range(1, n + 1):
    j = 1
    while j <= i:
        print(j,end="")
        j += 1
    print()
```
