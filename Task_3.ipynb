```python
# create a simple project where the computer thinks of a random number between 1 and 100, and the user has to guess it.
import random
n = random.randint(1,100)
a = int(input("Enter any number :"))
while n>0:
    if a!=n:
        print("wrong guess, please try again....")
        a = int(input("Enter another number :"))
    else :
        print("Correct guess")
        break
```
