```python
# Write a function that takes the radius of a circle and returns its area.
import math
def area(r):
    return 2 * math.pi * r
r = int(input("Enter the radius : "))
print(f'The area of the circle is {area(r)} sq units')
```

```python
# Write a function that accepts three numbers and returns the largest one.
def largest(a,b,c):
    if a>b and a>c:
        return a
    elif b>a and b>c:
        return b
    else :
        return c
a,b,c = map(int,input("Enter three numbers : ").split())
print(f'The largest number is {largest(a,b,c)}')
```

```python
# Write a recursive function to find the nth Fibonacci number.
def fib(n):
    if n==0:
        return 0
    if n==1:
        return 1
    return fib(n-2)+fib(n-1)
a=int(input("Enter which term is required: "))
print(fib(a))
```

```python
# Write a recursive function to reverse a string.
def reverse(word):
    return word[::-1]
word = input("Enter a string:")
print(f'The reverse of the word is {reverse(word)}')
```

```python
# Write a program to find the sum of all numbers from 1 to n using a for loop.
n=int(input("Enter the nth number :"))
sum = 0
for i in range(1,n+1):
    sum = sum + i
print(f'The sum of number from 1 to {n} is {sum}')
```

```python
# Write a program to count how many even numbers are present in a list.
lst = list(map(int,input("Enter the list :").split()))
total = 0
for i in lst:
    if i%2==0:
        total+=1
print(f'The number of even numbers are {total}')
```
